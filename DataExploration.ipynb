```python
import pandas as pd

df = pd.read_csv("data/case1/test1.csv")
df_test = pd.read_csv("data/case1/test1.csv")
```

```python
df["kilo_rpm"] = df["speed_rpm"] / 1000
```

```python
df["delta_p"] = df["outlet_pressure_bar"] - df["inlet_pressure_bar"]
```

```python
df
```

```python
df = df.drop(columns=['pump_id', 'operating_state'])
```

```python
df
```

```python
print(df.corr())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
correlation_matrix = df.corr()
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)

plt.title('Correlation Matrix Heatmap')
plt.show()
```

```python
dataset = df[["flow_m3_h"]].copy()
dataset['power_w**1/3'] = df['power_w']**(1/3)
dataset['power_w**1/2'] = df['power_w']**(1/2)
dataset['power_w'] = df['power_w']
dataset['power_w**2'] = df['power_w']**2
dataset['power_w**3'] = df['power_w']**3
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
correlation_matrix = dataset.corr()
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)

plt.title('Correlation Matrix Heatmap')
plt.show()
```

```python
dataset = df[["flow_m3_h"]].copy()
dataset['kilo_rpm**1/3'] = df['kilo_rpm']**(1/3)
dataset['kilo_rpm**1/2'] = df['kilo_rpm']**(1/2)
dataset['kilo_rpm'] = df['kilo_rpm']
dataset['kilo_rpm**2'] = df['kilo_rpm']**2
dataset['kilo_rpm**3'] = df['kilo_rpm']**3
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
correlation_matrix = dataset.corr()
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)

plt.title('Correlation Matrix Heatmap')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
correlation_matrix = dataset.corr(method='spearman')
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)

plt.title('Correlation Matrix Heatmap')
plt.show()
```
