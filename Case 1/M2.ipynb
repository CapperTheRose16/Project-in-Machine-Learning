```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

def relu(z):
    return np.maximum(0, z)
def sigmoid(z):
    x = 1/(1 + np.exp(-z))
    return x
def linear(z):
    return z 

ACTIVATIONS = {"relu": relu, "sigmoid": sigmoid, "linear": linear}

class NeuralNetwork:
    # I need to initialize the count of neurons and layers. 
    def __init__(self, input_size, hidden_layers, output_size, activations=None):
        # Initial checks
        if not isinstance(input_size, int) or input_size < 1:
                raise ValueError(f"input_size must be a positive integer, got {input_size}")
        if not isinstance(hidden_layers, list):
                raise TypeError(f"hidden_layers must be a list, got {type(hidden_layers)}")
        for i, n in enumerate(hidden_layers):
                if not isinstance(n, int) or n < 1:
                    raise ValueError(f"hidden_layers[{i}] must be a positive integer, got {n}")
        if not isinstance(output_size, int) or output_size < 1:
            raise ValueError(f"output_size must be a positive integer, got {output_size}")

        # Setting values
        self.input_size = input_size
        self.hidden_layers = hidden_layers
        self.output_size = output_size
        self.sizes = [input_size] + hidden_layers + [output_size]
        self.num_layers = len(self.sizes)

        n_transitions = len(self.sizes) - 1
        if activations is None:
            activations = ["relu"] * (n_transitions - 1) + ["linear"]
        if len(activations) != n_transitions:
            raise ValueError(f"need {n_transitions} activations, got {len(activations)}")
        for name in activations:
            if name not in ACTIVATIONS:
                raise ValueError(f"unknown activation {name}, choose from {list(ACTIVATIONS)}")
        self.activations = activations
    
        self.weights=[]
        self.biases=[]

        # Initiating every weight randomly
        # -1 everything execept last layer, 1 takes everything except the first layer. 
        for n_in, n_out in zip(self.sizes[:-1], self.sizes[1:]):
            W = np.random.randn(n_out, n_in) * np.sqrt(2 / n_in)   # He init
            b = np.zeros((n_out, 1))
            self.weights.append(W)
            self.biases.append(b)

    # I need a function to do feed forward. 
    def feedforward(self, a):
        pass
    
    # I need a function that chooses a minibatch to do gradient descent over. Maybe it can just randomly select index?
    def mini_batch(self):
        pass

    def update_mini_batch(self, mini_batch, eta):
        pass
    def backprop(self, x, y):
        pass

    def computing_loss(self):
        pass
```
