```python
import pandas as pd
import numpy as np
from DecisionTreeRegresion import DecisionTreeRegresion

features = ["power_w", "kilo_rpm", "delta_p"]
target = "flow_m3_h"

train_df = pd.read_csv("data/case1/train1.csv")
test_df = pd.read_csv("data/case1/test1.csv")

for df in (train_df, test_df):
    df["kilo_rpm"] = df["speed_rpm"] / 1000
    df["delta_p"] = df["outlet_pressure_bar"] - df["inlet_pressure_bar"]

model = DecisionTreeRegresion(max_depth=10)
model.fit(train_df[features], train_df[target])

predictions = model.predict(test_df[features])
model.print_tree()

test_mse = np.mean((test_df[target].to_numpy() - predictions) ** 2)
print(f"Test MSE: {test_mse:.3f}")
```
